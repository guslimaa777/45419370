## 3. Identificação de colunas numéricas e categóricas


```python

colunas_numericas = df.select_dtypes(include=['number']).columns.tolist()
colunas_categoricas = df.select_dtypes(include=['object', 'category']).columns.tolist()

print('Colunas numéricas:', colunas_numericas)
print('\nColunas categóricas:', colunas_categoricas)
```

## 4. Separação da variável-alvo

Utilize como variável-alvo a coluna `status_instrumentacao`.


```python

alvo = 'status_instrumentacao'

X = df.drop(columns=[alvo]).copy()
y = df[alvo].copy()

print('Dimensão de X:', X.shape)
print('Dimensão de y:', y.shape)

display(
    y.value_counts(dropna=False)
     .rename_axis('classe')
     .reset_index(name='quantidade')
)
```

## 5. Separação entre dados numéricos e categóricos


```python
X_numerico = X.select_dtypes(include=['int64', 'float64']).copy()
X_categorico = X.select_dtypes(include=["object", "category"]).copy()

print('dimensao de x_numerico:', X_numerico.shape)
print('dimensao de x_categorico:', X_categorico.shape)

display(X_categorico.head())
display(X_numerico.head())
```

## 6. Normalização

Complete o código para aplicar `MinMaxScaler` nas variáveis numéricas.


```python
scaler_minmax = MinMaxScaler()

X_normalizado_array = scaler_minmax.fit_transform(X_numerico)

X_normalizado = pd.DataFrame(
    X_normalizado_array,
    columns=X_numerico.columns,
    index=X_numerico.index
)

X_normalizado.head()
```

## 7. Comparação antes e depois da normalização


```python
comparacao_normalizacao = pd.DataFrame({
    'valor_original_min': [X_numerico['setpoint_vazao_l_min'].min()],
    'valor_original_max': [X_numerico['setpoint_vazao_l_min'].max()],
    'valor_normalizado_min': [X_normalizado['setpoint_vazao_l_min'].min()],
    'valor_normalizado_max': [X_normalizado['setpoint_vazao_l_min'].max()]
})

display(comparacao_normalizacao)
```

O que mudou nos dados após a normalização?

## 8. Padronização

Agora aplique `StandardScaler` às variáveis numéricas.


```python

scaler_standard = StandardScaler()

X_padronizado_array = scaler_standard.fit_transform(X_numerico)

X_padronizado = pd.DataFrame(
    X_padronizado_array,
    columns=X_numerico.columns,
    index=X_numerico.index
)

X_padronizado.head()
```

## 9. Comparação antes e depois da padronização


```python

comparacao_padronizacao = pd.DataFrame({
    'media_original': X_numerico.mean(),
    'desvio_original': X_numerico.std(),
    'media_padronizada': X_padronizado.mean(),
    'desvio_padronizado': X_padronizado.std()
})

display(comparacao_padronizacao.round(4))
```

## 10. Codificação de variáveis categóricas

Utilize `pd.get_dummies()` para transformar as variáveis categóricas em colunas numéricas.


```python
X_categorico_codificado = pd.get_dummies(X_categorico, drop_first=False)
X_categorico['timestamp']=pd.to_datetime(X_categorico['timestamp'])
print('dimensao da base categorica', X_categorico.nunique())
print('Dimensão da base categórica codificada:', X_categorico_codificado.shape)
X_categorico_codificado.head()


```

## 11. Preparação da base final para modelagem

Monte a base final usando:
- variáveis numéricas padronizadas;
- variáveis categóricas codificadas.


```python
X_final = pd.concat([X_padronizado, X_categorico_codificado], axis=1)

print('Dimensão da base final de atributos:', X_final.shape)
display(X_final.head())
```

## 12. Consolidação com a variável-alvo


```python
base_final_modelagem = X_final.copy()
base_final_modelagem[alvo] = y

print('Dimensão da base final completa:', base_final_modelagem.shape)
display(base_final_modelagem.head())
```

## 13. Verificações finais


```python
display(
    base_final_modelagem.isnull()
    .sum()
    .rename_axis('coluna')
    .reset_index(name='faltantes')
)

display(
    base_final_modelagem.dtypes
    .rename_axis('coluna')
    .reset_index(name='tipo')
)
```
